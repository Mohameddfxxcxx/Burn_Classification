```python
%%writefile app.py

import streamlit as st
import tensorflow as tf
from tensorflow.keras.preprocessing import image
import numpy as np
from tensorflow.keras.applications.efficientnet import preprocess_input
from PIL import Image

# Set page configuration (MUST BE THE FIRST STREAMLIT COMMAND)
st.set_page_config(
    page_title="Burn Degree Classification",
    page_icon="🔥",
    layout="wide",
    initial_sidebar_state="expanded"
)

# Custom DepthwiseConv2D class to handle loading the model
class CustomDepthwiseConv2D(tf.keras.layers.DepthwiseConv2D):
    def __init__(self, *args, **kwargs):
        kwargs.pop('groups', None)
        super().__init__(*args, **kwargs)

# Load the model using st.cache_resource
@st.cache_resource
def load_model():
    model = tf.keras.models.load_model(
        '/content/drive/MyDrive/my_model_eff2.h5',
        custom_objects={'DepthwiseConv2D': CustomDepthwiseConv2D}
    )
    return model

model = load_model()

# Class names
class_names = ['First Degree Burn', 'Second Degree Burn', 'Third Degree Burn']

# Custom CSS for background and styling
st.markdown(
    """
    <style>
    .stApp {
        background-color: #8B4513; /* SaddleBrown color for the main background */
        color: white; /* White text color for better contrast */
    }
    .stButton>button {
        background-color: #ff4b4b;
        color: white;
        border-radius: 10px;
        padding: 10px 20px;
        font-size: 16px;
    }
    .stMarkdown {
        color: white; /* Ensure all text is white */
    }
    .stSidebar {
        background-color: rgba(255, 255, 255, 0.9); /* Light background for sidebar */
        padding: 20px;
        border-radius: 10px;
        color: black; /* Black text for sidebar */
    }
    .stSidebar h1, .stSidebar h2, .stSidebar h3, .stSidebar h4, .stSidebar h5, .stSidebar h6 {
        color: black; /* Ensure headers are black */
    }
    .stSidebar p, .stSidebar li {
        color: black; /* Ensure paragraphs and list items are black */
    }
    </style>
    """,
    unsafe_allow_html=True
)

# Sidebar for additional information
with st.sidebar:
    st.title("🔥 Burn Treatment Guide")
    st.write("""
    **First Degree Burn:**
    - Cool the burn with cool (not cold) water.
    - Apply aloe vera or moisturizer.
    - Avoid ice or butter.

    **Second Degree Burn:**
    - Cool the burn with cool water for 15-20 minutes.
    - Apply antibiotic ointment.
    - Cover with a non-stick bandage.

    **Third Degree Burn:**
    - Seek immediate medical attention.
    - Do not attempt to treat at home.
    - Cover the burn with a clean, dry cloth.
    """)

# Main content
st.title("Burn Degree Classification")
st.write("Upload an image of a burn to classify its degree.")

uploaded_file = st.file_uploader("Choose an image...", type=["jpg", "jpeg", "png"])

if uploaded_file is not None:
    # Display the uploaded image
    image_display = Image.open(uploaded_file)
    st.image(image_display, caption='Uploaded Image', use_column_width=True)

    # Preprocess the image for prediction
    img = image_display.resize((224, 224))
    img_array = image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0)
    img_array = preprocess_input(img_array)

    # Make prediction
    prediction = model.predict(img_array)
    predicted_class = np.argmax(prediction)

    # Display the prediction
    st.markdown(f"**Prediction:** {class_names[predicted_class]}")
    st.markdown(f"**Confidence:** {np.max(prediction) * 100:.2f}%")

    # Additional information based on prediction
    if predicted_class == 0:
        st.info("**First Degree Burn:** This is a mild burn affecting only the outer layer of the skin. Follow the treatment guide in the sidebar for more information.")
    elif predicted_class == 1:
        st.warning("**Second Degree Burn:** This burn affects both the outer and underlying layer of the skin. Follow the treatment guide in the sidebar for more information.")
    elif predicted_class == 2:
        st.error("**Third Degree Burn:** This is a severe burn that requires immediate medical attention. Follow the treatment guide in the sidebar for more information.")

# Footer
st.markdown("---")
st.markdown("**Disclaimer:** This app is for informational purposes only and is not a substitute for professional medical advice.")
```
